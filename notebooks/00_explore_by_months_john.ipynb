```python
def check_month(path):
    df = pd.read_csv(path)
    print(path)
    print('Shape:', df.shape)
    print('Missing values:')
    print(df.isnull().sum())
    print('Trips under 60 sec:', (df['Trip_Duration'] < 60).sum())
    print('Trips over 6 hours:', (df['Trip_Duration'] > 21600).sum())
    print('Same start/end station:', (df['Start_Station_Id'] == df['End_Station_Id']).sum())
    print('Date range:', df['Start_Time'].min(), df['Start_Time'].max())
```

```python
check_month(f'{RAW}/2025/bikeshare_2025_01.csv')
```

```python
for m in range(1, 13):
    month = str(m).zfill(2)
    check_month(f'{RAW}/2025/bikeshare_2025_{month}.csv')
    print('-' * 40)
```

```python
rows = []
for m in range(1, 13):
    month = str(m).zfill(2)
    df = pd.read_csv(f'{RAW}/2025/bikeshare_2025_{month}.csv')
    rows.append({
        'month': month,
        'trips': len(df),
        'missing_end_id': df['End_Station_Id'].isnull().sum(),
        'missing_end_time': df['End_Time'].isnull().sum(),
        'under_60s': (df['Trip_Duration'] < 60).sum(),
        'over_6h': (df['Trip_Duration'] > 21600).sum(),
        'same_station': (df['Start_Station_Id'] == df['End_Station_Id']).sum(),
        'first_trip': df['Start_Time'].min(),
        'last_trip': df['Start_Time'].max(),
    })
summary_2025 = pd.DataFrame(rows)
summary_2025
```

2025 summary (all 12 months):

1. The month with the most trips under 60 seconds is October (2,911 trips). It is not the busiest month.
2. Missing values are only in End_Station_Id and End_Time. In September (7), November (528) and December (389), End_Station_Name is also missing.
3. All dates stay inside their own month. Nothing removed.
